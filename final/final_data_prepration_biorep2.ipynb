# Biorep2 Data Preparation (Independent Batch Validation)

**What this notebook does:**  
This notebook prepares the **biorep2** TD–GC–MS peak tables into analysis-ready datasets. It reads QC and sample inputs via **`config.yaml`**, cleans and standardizes the tables, evaluates QC stability, performs **LOESS-based drift correction** using injection order, applies **QC-based normalisation (QC1)**, and attaches **Variety** metadata before exporting final outputs.

**How it is used (independent batch):**  
The resulting biorep2 outputs are used as an **independent batch** to evaluate whether models, signals, and preprocessing decisions derived from **biorep1** generalise to a separate analytical run (i.e., batch validation beyond internal cross-validation).

**Main steps (high level):**
- Load **biorep2 QC** section from the QC Excel file (paths/settings come from `config.yaml`)
- Clean and standardize identifiers/columns (retention time, m/z, sample/QC naming)
- Merge QC data with master metadata to obtain **injection order**
- QC checks: stability filtering and linearity assessment
- **Drift correction**: LOESS smoothing across injection order
- **Normalisation**: QC-based scaling (QC1) on corrected data
- Attach **Variety** info to sample metadata and save final tables
- Beacuse biorep2 is just for model evaluation and every step is like biorep1, I did not add any explanation here.

**Module dependency:**  
This notebook uses reusable helper functions from **`data_preparation.py`** (implemented in this repo as **`datapreparation_module.py`**).
